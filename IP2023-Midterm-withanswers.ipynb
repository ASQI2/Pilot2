## Question 1 [20 points]

Define a function called password() that accepts one (string) argument: a password. 

Your function should perform a series of checks, to decide whether the password is valid or not. 

A password is considered valid if **all** of the following statements are satisfied:

- The password's length is equal to or greater than 8 characters
- The password contains at least one valid upper-case character (valid upper-case characters: ABCDEFGHIJKLMNOPQRSTUVWXYZ)
- The password contains at least one valid number character (valid number characters: 0123456789)
- The password contains at least one valid special character (valid special characters: !@#$%^&*)

Your function should **return** True (as a boolean) when the password given as argument to the function satisfies all the above statements. It should **return** False (as a boolean) otherwise.

```python
def password(s):
    # your code here
    uppercase = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    nums = '1234567890'
    special = '!@#$%^&*'
    
    containsUpper = False
    containsNum = False
    containsSpecial = False
    
    if len(s) < 8:
        return False
    
    for char in s:
        if char in uppercase:
            containsUpper = True
        elif char in nums:
            containsNum = True
        elif char in special:
            containsSpecial = True
    
    return containsUpper and containsNum and containsSpecial
#---------------------------------------------------------------------------------------------------------------------------
if __name__ == '__main__':
    print(password("ThisPasswordIsValid!!!")) # Expected output: False
    print(password("totallYwrongpassword1234@@@")) # Expected output: True
    print(password("hello!2")) # Expected output: False
    print(password("python1234!@$#")) # Expected output: False
    print(password('hello12346')) # Expected output: False
    print(password('SMALL')) # Expected output: False
    print(password('123456789')) # Expected output: False
    print(password('&@!#^$*&^!')) # Expected output: False
    
```

## Question 2 [20 points]

Define a function called distance3D() that accepts two tuples as arguments. 

Each of the two tuples should contain three numbers, which are coordinates of a point in 3D space.

Your function should calculate and return the euclidean distance between the two points as a float, **rounded to 1 decimal**.

You can assume that both tuples will always contain three valid numbers. You can assume that the tuples will always follow the (x, y, z) format (x-coordinate, followed by y-coordinate, followed by z-coordinate).

Euclidean distance between points in 3D space:

$Euclidean_{3D} = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2 + (z_2-z_1)^2}$ 

where $x_1, y_1, z_1$ refer to the first point (first argument tuple) and $x_2, y_2, z_2$ refer to the second point (second argument tuple).


```python
import math

def distance3D(t1,t2):
    # your code here
    return round(math.sqrt((t2[0]-t1[0])**2 + (t2[1]-t1[1])**2 + (t2[2]-t1[2])**2),1)

#---------------------------------------------------------------------------------------------------------------------------   
if __name__ == '__main__':
    print(distance3D((1,1,1),(1,1,1))) # Expected output: 0.0
    print(distance3D((4,4,4),(1,1,1))) # Expected output: 5.2
    print(distance3D((10,20,30),(9,19,29))) # Expected output: 1.7 
    print(distance3D((1,2,3),(-1,-2,-3))) # Expected output: 7.5 
    
```

## Question 3 [20 points]

Define a function called letterList() that accepts one argument (a string). 

Your function should return a list of all valid letters of the English alphabet that are present in the argument string, **sorted from a to z**. Each letter should only be added to the list once (for example, if the letter 'b' is found twice (or more times), only add it to the list once). 

Valid English letters: abcdefghijklmnopqrstuvwxyz (and the equivalent upper-case: ABCDEFGHIJKLMNOPQRSTUVWXYZ)

Your function should be **case-insensitive** (count both upper- and lower-case characters as the same letter). **The characters in the returned list should be all lower-case.**



```python
def letterList(s):
    # your code here
    letters = 'abcdefghijklmnopqrstuvwxyz'
    res = []
    
    for char in s:
        if char.lower() in letters and not char.lower() in res:
            res.append(char.lower())
    res.sort()
    return res
    
    

if __name__ == '__main__':
    print(letterList("TrUsT mE, I aM aN eNgInEeR!")) # # Expected output: ['a', 'e', 'g', 'i', 'm', 'n', 'r', 's', 't', 'u']
    print(letterList("!@#%$^%$^&&&^%&^%@$#@$#@!#@$#")) # # Expected output: []
    print(letterList("this exam is going great")) # # Expected output: ['a', 'e', 'g', 'h', 'i', 'm', 'n', 'o', 'r', 's', 't', 'x']
    print(letterList("WHY ARE YOU SHOUTING AT ME")) # # Expected output: ['a', 'e', 'g', 'h', 'i', 'm', 'n', 'o', 'r', 's', 't', 'u', 'w', 'y']
    print(letterList("ZXCVBNMASDFGHJKLQWERTYUIOP")) # # Expected output: ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']
    print(letterList("1234567890")) # # Expected output: []
```

## Question 4 [20 points]

My cat is supposed to eat at 7 in the morning. For every hour that passes while she has not been fed, she doubles the amount of meows she makes, starting at 1 meow at 7 o'clock. So, at 8, she will meow twice, four meows at 9, eight meows at 10, etc.

Define a function called meow() which accepts one argument: an integer indicating what hour of the day it currently is. Your function should return an integer, indicating the total amount of meows the cat will make. Note: do not return the amount of meows she will make at the given time, but the **total** amount of meows up to and including the given time, since 7 o'clock. 

If the argument integer is below 7 or above 23, your function should return False (as a boolean).

If the argument integer is between (and including) 7 and 23, calculate and return (as an integer) the appropriate amount of meows.

You can assume the argument will always be an integer.

```python
def meow(time):
    # your code here
    if time < 7 or time > 23:
        return False
    
    res = 0
    meows = 1
    for i in range(7,time+1):
        res += meows
        meows*=2
    return res
    
    
#--------------------------------------------------------------------------------------------------------------------------------------------------
if __name__ == '__main__':
    print(meow(10)) # Expected output: 15
    print(meow(1)) # Expected output: False
    print(meow(7)) # Expected output: 1
    print(meow(15)) # Expected output: 511
    print(meow(23)) # Expected output: 131071
    print(meow(-8)) # Expected output: False
    
```

## Question 5 [20 points]

Define a function called characterCount() that accepts one argument: a string.

The argument string can be a phrase of any length containing any character. 

Your function should return a list of tuples, counting how many times each unique character appears in the argument string.

Each tuple should contain two elements, the first one being a character (string) and the second one being the number of appearances (integer) of that particular character in the argument string.

The returned list of tuples should only consider characters that appear in the string. Each tuple should consider a different (unique) character (each unique character should only be considered once). Your function should consider lower-case and upper-case letters as different characters (case-sensitive). Your function should count all possible characters (including spaces, number characters and other special characters).

The ordering of the tuples in the returned list does not matter.

```python
def characterCount(s):
    # your code here
    res = []
    checked = []
    for char in s:
        if not char in checked:
            res.append((char, s.count(char)))
            checked.append(char)
    return res
    
    
#------------------------------------------------------------------------------------------------------------------------------------------
if __name__ == '__main__':
    print(characterCount('hiiiiiiiiiiii!')) # Expected output: [('h', 1), ('i', 12), ('!', 1)]
    print(characterCount('What an "Amazing" exercise...')) # Expected output: [('W', 1), ('h', 1), ('a', 3), ('t', 1), (' ', 3), ('n', 2), ('"', 2), ('A', 1), ('m', 1), ('z', 1), ('i', 2), ('g', 1), ('e', 3), ('x', 1), ('r', 1), ('c', 1), ('s', 1), ('.', 3)]
    print(characterCount('this text ------ is    345677767        weird!@#xD')) # Expected output: [('t', 3), ('h', 1), ('i', 3), ('s', 2), (' ', 15), ('e', 2), ('x', 2), ('-', 6), ('3', 1), ('4', 1), ('5', 1), ('6', 2), ('7', 4), ('w', 1), ('r', 1), ('d', 1), ('!', 1), ('@', 1), ('#', 1), ('D', 1)]
    print(characterCount('HeLlo123321!@#')) # Expected output: [('H', 1), ('e', 1), ('L', 1), ('l', 1), ('o', 1), ('1', 2), ('2', 2), ('3', 2), ('!', 1), ('@', 1), ('#', 1)]
    print(characterCount('aaaAAAbbbBBB')) # Expected output: [('a', 3), ('A', 3), ('b', 3), ('B', 3)]
```
